```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# 1. Loading the Data
# Load the Iris dataset directly from the Scikit-learn library
# This contains 150 samples of iris flowers with 4 features
# and their corresponding species labels

iris = datasets.load_iris()


# 2. Creating the Dataframes
# Convert the raw array data into a Pandas DataFrame
# We also map the numeric targets to their actual names
# so the data is human-readable and easier to inspect

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target
df['species_name'] = df['species'].apply(lambda x: iris.target_names[x])


# 3. Visualization of Data
# Create a "pairplot" to visualize how different features relate to each other
# Different colors represent different species; if the colors are separated nicely,
# it means the SVM will likely have an easy time classifying them.

sns.pairplot(df, hue='species_name', palette='brg')
plt.show()


# 4. The training data
# Separate the features from the target
# We drop the helper columns we created for visualization
# so the model only sees the raw measurement numbers.

X = df.drop(['species', 'species_name'], axis=1)
y = df['species']


# 5. Splitting the data
# Split the dataset into two pieces: 70% for training and 30% for testing.
# This ensures we have "unseen" data to test how well the model generalizes
# instead of just memorizing the training examples.

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)


# 6. Model Intializiation
# Create the Support Vector Machine classifier with a linear kernel.
# A 'linear' kernel tries to draw straight lines to separate
# the different flower species clusters

model = SVC(kernel='linear')


# 7. Training the Model
# Feed the training data which is the X_train and answers,
# which is the y_train into the model
# The algorithm now adjusts its internal math to find the best
# boundary lines that separate the three species.

print("Training the model...")
model.fit(X_train, y_train)


# 8. Model Visualization Performance
# Ask the model to predict species for the test data (X_test) it hasn't seen yet.
# Then, compare those predictions, the y_pred to the actual answers, the y_test
# to calculate the accuracy percentage.

print("Evaluating...")
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=iris.target_names))


"""
REFERENCES:
1. Scikit-learn SVM Documentation:
   https://scikit-learn.org/stable/modules/svm.html

2. The Iris Dataset (UCI Machine Learning Repository):
   https://archive.ics.uci.edu/ml/datasets/iris

3. Scikit-learn 'load_iris' Utility:
   https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_iris.html

4. Pandas DataFrame Documentation (for data manipulation):
   https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html

5. Seaborn Pairplot Documentation (for visualization):
   https://seaborn.pydata.org/generated/seaborn.pairplot.html
"""
```
